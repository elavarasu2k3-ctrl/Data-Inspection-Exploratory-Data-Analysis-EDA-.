# Notebook 13 - EDA Report

## 1. Concept

The EDA report is a summary document that pulls together everything discovered across Notebooks 1-12 into one organized write-up. Instead of scattered findings across many notebooks, this notebook presents the dataset overview, data quality issues, statistical findings, relationships, visual insights, and recommendations in one place.

This is the kind of document that gets handed to a manager, a data science lead, or the next engineer who picks up the project - so it needs to be readable without re-running any code.

## 2. Example

A short version of what a real EDA report section looks like:

"The dataset contains 5,020 customer records with 14 features related to telecom subscriptions. About 6% of rows had missing values in `TotalCharges`, and 2% were exact duplicates. The target variable `Churn` is imbalanced, with about 27% of customers churning. Contract type and tenure showed the strongest relationship with churn."

This is different from a notebook-by-notebook dump - it's a condensed, decision-ready summary.

```python
import pandas as pd

df = pd.read_csv("custumer.csv")

# Dataset overview
print("Shape:", df.shape)
print("\nColumns:", list(df.columns))
print("\nData types:\n", df.dtypes)

# Data quality summary
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
print("\nMissing values:\n", missing[missing > 0])
print("\nMissing % :\n", missing_pct[missing_pct > 0])

duplicate_count = df.duplicated().sum()
print("\nDuplicate rows:", duplicate_count)

# Target variable summary
churn_counts = df["Churn"].value_counts()
churn_pct = df["Churn"].value_counts(normalize=True) * 100
print("\nChurn counts:\n", churn_counts)
print("\nChurn %:\n", churn_pct)

# Key correlations (numerical features only)
numeric_df = df.select_dtypes(include=["int64", "float64"])
correlation = numeric_df.corr()
print("\nCorrelation matrix:\n", correlation)
```
