### What is Data Inspection?

Data inspection is the process of examining the `custumer.csv` dataset to understand its structure, contents, data types, distributions, and potential problems before performing data analysis or machine learning.

Real-world example:

A telecom company receives customer records via `custumer.csv` containing:

* Customer ID
* Age
* Contract type
* Monthly charges
* Internet service
* Churn status

Before building a churn prediction model, we need to inspect the data and answer questions such as:

* How many customers are there?
* Which columns are numerical?
* Which are categorical?
* Are there missing values?
* What values occur in each categorical column?

```python
import pandas as pd

df = pd.read_csv("custumer.csv")
df.head()
```

### 1. Number of Rows

The number of rows represents the number of customer records or observations in `custumer.csv`.

In this dataset, each row represents one distinct customer.

Real-world example:

If a telecom company maintains records for 5,020 customers, `custumer.csv` will contain 5,020 rows.

#### AI/ML usage

The number of observations affects:

* Model training
* Model accuracy
* Statistical reliability
* Train/test splitting

More useful observations generally give an ML model more information to learn from.

```python
print("Number of rows in custumer.csv:", len(df))
```

### 2. Number of Columns

The number of columns represents the number of variables or customer attributes present in `custumer.csv`.

Real-world example:

For each customer in `custumer.csv`, we track:

* Age
* Gender
* Monthly charges
* Contract type
* Internet service
* Churn

Each attribute is mapped to a single column.

#### AI/ML usage

Columns usually become features used by an ML model.

```python
print("Number of columns in custumer.csv:", len(df.columns))
```

```python
print("Available columns:", df.columns.tolist())
```

### 3. DataTypes

A data type defines how values are stored and interpreted in each column of `custumer.csv`.

Common types include:

| Data Type  | Meaning                       | Example      |
| ---------- | ----------------------------- | ------------ |
| `int64`    | Integer                       | 5, 10, 20    |
| `float64`  | Decimal number                | 25.5, 100.75 |
| `object`   | Usually text/categorical data | Male, Female |
| `datetime` | Date/time                     | 2026-08-24   |

#### AI/ML usage

Data types determine what preprocessing steps (such as encoding or scaling) are required for `custumer.csv`.

```python
df.dtypes
```

By default, Pandas assigns an integer RangeIndex starting from 0.

Real-world example:

The index acts as an internal row reference:

Index     CustomerID
0         CUST00001
1         CUST00002
2         CUST00003

#### AI/ML usage

The index is positional metadata and is not used as a predictive feature for churn classification.

```python
print(df.index)
```

### 16. Feature Engineering: Grouping Continuous Data

Binning `TenureYears` in `custumer.csv` creates a structured categorical feature `TenureGroup`.

```python
df["TenureGroup"] = pd.cut(
    df["TenureYears"],
    bins=[0, 2, 5, 10, float("inf")],
    labels=["0-2 Years", "3-5 Years", "6-10 Years", "10+ Years"],
)

df[["CustomerID", "TenureYears", "TenureGroup"]].head()
```
